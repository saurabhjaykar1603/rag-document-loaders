# Document Loading
Run each example independently. Local sources use the generic files in `data/`.

## 01. Text Loader

```python
from pathlib import Path

DATA_DIR = Path.cwd() / "data"

from langchain_community.document_loaders import TextLoader

loader = TextLoader(str(DATA_DIR / "rag_notes.txt"), encoding="utf-8")
docs = loader.load()

print("Documents:", len(docs))
print(docs[0].metadata)
print(docs[0].page_content)
```

## 02. Pdf Loader

```python
from pathlib import Path

DATA_DIR = Path.cwd() / "data"

from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader(str(DATA_DIR / "document_loading_guide.pdf"))
docs = loader.load()

print("Pages:", len(docs))
for doc in docs:
    print("\nMetadata:", doc.metadata)
    print(doc.page_content)
```

## 03. Csv Loader

```python
from pathlib import Path

DATA_DIR = Path.cwd() / "data"

from langchain_community.document_loaders import CSVLoader

loader = CSVLoader(str(DATA_DIR / "products.csv"), encoding="utf-8")
docs = loader.load()

print("Rows:", len(docs))
for doc in docs:
    print("\nMetadata:", doc.metadata)
    print(doc.page_content)
```

## 04. Web Loader

```python
import os

os.environ.setdefault("USER_AGENT", "RAGLoaderExamples/1.0 (https://github.com/saurabhjaykar1603/rag-document-loaders)")

from langchain_community.document_loaders import WebBaseLoader

loader = WebBaseLoader(
    web_path="https://www.example.com",
    requests_kwargs={"timeout": 30},
)
docs = loader.load()

print("Documents:", len(docs))
for doc in docs:
    print(doc.metadata)
    print(doc.page_content)
```

## 05. Wikipedia Loader

```python
from langchain_community.document_loaders import WikipediaLoader

loader = WikipediaLoader(
    query="Python (programming language)",
    load_max_docs=3,
    doc_content_chars_max=12000,
)
docs = loader.load()

print("Articles:", len(docs))
for doc in docs:
    print("\nMetadata:", doc.metadata)
    print(doc.page_content)
```
